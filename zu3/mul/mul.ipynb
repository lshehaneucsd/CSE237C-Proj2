# How to start board

- Plug in board power
- Plug in Computer to board
- Turn on board
```PS
Get-PnpDevice -PresentOnly | Where-Object {$_.InstanceId -like 'USB*'} | Select-Object Status, FriendlyName
```
should show: `OK     Linux File-Stor Gadget USB Device`

Wait for all the LEDs to shine (it'll go blue and purple)

Go to `http://192.168.3.1:9090/lab`, password: `xilinx`
