## Value Maker
This is for making the angles and kvalues to update the cordic arrays

```python
import math

NO_ITER = 20

def fmt_sig(val, sig_figs=15):
    if val == 1:
        return "1"
    if val < 1e-4:
        return f"{val:.{sig_figs-1}e}"
    leading_zeros = -int(math.floor(math.log10(val))) - 1
    return f"{val:.{leading_zeros + sig_figs}f}"

angles = [fmt_sig(math.atan(2**(-i)), 15) for i in range(NO_ITER)]
kvalues = [fmt_sig(2.0**(-i), 15) for i in range(NO_ITER)]

print(f"data_t Kvalues[NO_ITER] = {{{', '.join(kvalues)}}};\n")
print(f"data_t angles[NO_ITER] = {{{', '.join(angles)}}};\n")
```
